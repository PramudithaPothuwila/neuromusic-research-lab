# Results analysis
Loads persisted experiments. Held-out metrics only; training accuracy is shown for reference.

```python
import pandas as pd
from neuromusic.workspace import Workspace
from neuromusic.experiments.persistence import ExperimentStore
ws = Workspace()
store = ExperimentStore(ws.experiments_dir)
pd.DataFrame([{'id': r.experiment_id, 'type': r.experiment_type.value, 'status': r.status.value, 'participants': r.n_participants} for r in store.list_records()])
```

```python
clf = store.list_records('classification')[0]
pd.DataFrame(store.load_json(clf.experiment_id, 'metrics')).T[['accuracy', 'balanced_accuracy', 'macro_f1', 'mean_train_accuracy']]
```

```python
stats = store.list_records('statistics')[0]
store.load_table(stats.experiment_id, 'comparisons').head(20)
```
