# Exploratory analysis
Uses the `neuromusic` package; computation lives in the library, not here. Run `neuromusic scan` and `neuromusic features` first.

```python
from neuromusic.workspace import Workspace
from neuromusic import services
ws = Workspace()
manifest = ws.registry.manifest_frame(services.song_titles(ws))
manifest['provenance'].value_counts()
```

```python
manifest[manifest['analysable']].groupby('song_title').size().sort_values()
```

```python
rec_id = ws.registry.analysable_recordings()[0].recording_id
res = services.load_preprocessed(ws, rec_id)
res.channel_quality
```

```python
res.flags.head(20)
```

```python
trials = services.load_features(ws)
trials.filter(like='region.rel_power.alpha').describe()
```
